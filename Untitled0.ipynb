```python
import numpy as np
import threading
import time

from flask import Flask, render_template_string
from google.colab import output

print("Libraries imported successfully!")
```

```python
# Five web pages
pages = ["A", "B", "C", "D", "E"]

# Links between the pages
links = {
    "A": ["B", "C"],
    "B": ["C", "D"],
    "C": ["A", "E"],
    "D": ["C", "E"],
    "E": ["A", "D"]
}

print("Web pages:", pages)

print("\nLinks:")
for page, destinations in links.items():
    print(f"{page} -> {', '.join(destinations)}")
```

```python
n = len(pages)

# Create an n x n matrix filled with zeros
link_matrix = np.zeros((n, n), dtype=int)

# Fill the matrix
for i, page in enumerate(pages):
    for destination in links[page]:
        j = pages.index(destination)
        link_matrix[i, j] = 1

print("LINK MATRIX")
print("=" * 40)

print("     ", end="")
for page in pages:
    print(f"{page:^8}", end="")
print()

for i, page in enumerate(pages):
    print(f"{page:^5}", end="")
    for value in link_matrix[i]:
        print(f"{value:^8}", end="")
    print()
```

```python
# Transition matrix
P = np.zeros((n, n))

for i in range(n):

    number_of_links = np.sum(link_matrix[i])

    if number_of_links == 0:
        # If a page has no outgoing links
        P[i, :] = 1 / n
    else:
        P[i, :] = link_matrix[i] / number_of_links

print("TRANSITION PROBABILITY MATRIX")
print("=" * 50)

print("     ", end="")
for page in pages:
    print(f"{page:^10}", end="")
print()

for i, page in enumerate(pages):
    print(f"{page:^5}", end="")
    for value in P[i]:
        print(f"{value:^10.3f}", end="")
    print()
```

```python
# Damping factor
d = 0.85

# Number of pages
n = len(pages)

# Google PageRank matrix
G = d * P.T + ((1 - d) / n) * np.ones((n, n))

print("GOOGLE PAGERANK MATRIX")
print("=" * 55)

print("     ", end="")
for page in pages:
    print(f"{page:^10}", end="")
print()

for i, page in enumerate(pages):
    print(f"{page:^5}", end="")
    for value in G[i]:
        print(f"{value:^10.4f}", end="")
    print()

print("\nDamping factor =", d)
```

```python
# Initial PageRank vector
rank = np.ones(n) / n

# Store every iteration
iteration_history = []

# Maximum number of iterations
max_iterations = 100

# Convergence tolerance
tolerance = 1e-10

for iteration in range(max_iterations):

    new_rank = G @ rank

    iteration_history.append(new_rank.copy())

    # Check convergence
    difference = np.linalg.norm(new_rank - rank, 1)

    if difference < tolerance:
        rank = new_rank
        break

    rank = new_rank

# Normalize the final vector
rank = rank / np.sum(rank)

print("POWER METHOD")
print("=" * 50)

print("Number of iterations:", len(iteration_history))

print("\nFinal PageRank Vector:")

for i, page in enumerate(pages):
    print(f"Page {page}: {rank[i]:.6f}")
```

```python
# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(G)

# Find eigenvalue closest to 1
index = np.argmin(np.abs(eigenvalues - 1))

dominant_eigenvalue = eigenvalues[index].real
eigenvector = eigenvectors[:, index].real

# Make eigenvector positive
if np.sum(eigenvector) < 0:
    eigenvector = -eigenvector

# Normalize
eigenvector = eigenvector / np.sum(eigenvector)

print("EIGENVALUE AND EIGENVECTOR")
print("=" * 50)

print(f"Dominant Eigenvalue = {dominant_eigenvalue:.6f}")

print("\nCorresponding Eigenvector:")

for i, page in enumerate(pages):
    print(f"Page {page}: {eigenvector[i]:.6f}")
```

```python
print("COMPARISON")
print("=" * 65)

print(f"{'Page':<10}{'Power Method':<20}{'Eigenvector':<20}")

for i, page in enumerate(pages):
    print(
        f"{page:<10}"
        f"{rank[i]:<20.6f}"
        f"{eigenvector[i]:<20.6f}"
    )
```
